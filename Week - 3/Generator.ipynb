```python
# Generator is a function that bahaves like an iterator. (it can be used in a for loop).
# It pauses the function, return a value, and then resumes the function
# Used yeild instead of return.
# Iterate without loading everything into memory at once. (memory efficient, reading large files)
# return = pouring bucket
# yeild = drip faucet
```

```python
def topten():
    n = 1

    while n <= 10:
        sq = n*n
        yield sq
        n += 1

values = topten()

for i in values:
    print(i)
```

```python
def read_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()

file_path = 'test.txt'

for line in read_file(file_path):
    print(line)
```
