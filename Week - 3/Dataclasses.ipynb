```python
#without dataclasses
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"
    def __eq__(self, other):
        return self.name, self.age == other.name, other.age
        
p = Person("Alice", 30)
p1 = Person("Alice", 30)
print(p == p1)  # True
```

```python
#with dataclasses

from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int

p = Person("Alice", 30)
p.name, p.age
```

```python
p1 = Person("Alice", 30)
p == p1
```
