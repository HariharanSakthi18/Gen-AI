```python
# POC 1 — Login Required Decorator
# Interview Question
# Only logged-in users should create prompts.

templates = [
    {
        "id": 1,
        "name": "John",
        "category":"Marketing",
        "prompt":"Write a professional email..."
    },
]

logged_in = True

def login_required(func):
    def wrapper(*args, **kwargs):
        if logged_in:
            return func(*args, **kwargs)
        else:
            print("Login first..")
    return wrapper

@login_required
def original(id,name,category,prompt):
    templates.append(
        {
            "id":id,
            "name":name,
            "category":category,
            "prompt":prompt
        }
    )
    print("Prompt Added")

original(
    2,
    "Hari",
    "Software Developer",
    "Develop an Application.."
)

print(templates)
```
