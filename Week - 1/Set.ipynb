```python
# Sets are unordered collections of unique elements. They are immutable, Once a set is created, you cannot change its items, but you can add new items. Sets are defined using curly braces {} or the built-in set() function.
```

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
print(my_set) 
```

```python
new = {4,7,2,8,9}
print(new)
```

```python
# Adding an element to a set
my_set.add(6)
print(my_set)

# duplicates are not allowed in sets
my_set.add(3)  # This will not add another '3' to the set   
print(my_set)
```

```python
my_set.add(False)  # Adding a boolean value to the set
my_set.add(True)   # Adding another boolean value to the set
my_set.add(0)     # Adding an integer value to the set
my_set.add(1)     # Adding another integer value to the set
print(my_set)  # The set will only contain one instance of each unique value
```

```python
# updating a set with multiple elements
my_set.update([7, 8, 9])
print(my_set)
```

```python
print(len(my_set))  

print(type(my_set))
```

```python
# converting a set to a list
my_list_from_set = list(my_set)
print(my_list_from_set)

# converting a list to a set
my_set_from_list = set(my_list_from_set)
print(my_set_from_list)

#converting a set to a tuple
my_tuple_from_set = tuple(my_set)
print(my_tuple_from_set)
```

```python
# Removing an element from a set
my_set = {1, 2, 3, 4, 5}
my_set.remove(3)    
print(my_set)

s = {"apple", "banana", "cherry" , "carrot", "potato", "onion", "tomato"}
s.remove("banana")
print(s)

s.discard("onion")  # This will not raise an error even if "onion" is not in the set
print(s)

s.pop()
print(s)

s.clear()  # This will remove all elements from the set 's'
print(s)  # This will print an empty set
```

```python
# union of sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)   # You can use the | operator instead of the union() method, and you will get the same result.
print(union_set) 

# union multiple sets
set3 = {5, 6, 7}
union_set_multiple = set1 | set2 | set3  # you can use this method also set1.union(set2, set3) to get the same result
print(union_set_multiple)
```

```python
# intersection of sets
intersection_set = set1.intersection(set2)   # You can use the & operator instead of the intersection() method, and you will get the same result.
print(intersection_set)
```

```python
# difference of sets
difference_set = set1.difference(set2)   # You can use the - operator instead of the difference() method, and you will get the same result.
print(difference_set)   

# symmetric difference of sets
symmetric_difference_set = set1.symmetric_difference(set2)   # You can use the ^ operator instead of the symmetric_difference() method, and you will get the same result.
print(symmetric_difference_set)
```

```python
#isdisjoint() method returns True if two sets have no elements in common, otherwise it returns False.
set_a = {1, 2, 3}   
set_b = {4, 5, 6}
print(set_a.isdisjoint(set_b))
```

```python
# issubset() method returns True if all elements of the set are present in another set, otherwise it returns False.
set_c = {1, 2, 3, 4, 5}
print(set_a.issubset(set_c)) 
```

```python
# issuperset() method returns True if all elements of another set are present in the set, otherwise it returns False.
# If your logic requires you to start with the larger configuration set (c), you must use issuperset to get a True result
print(set_a.issuperset(set_c)) # is False (because a does not contain all of c).
```

```python
# POC 1. Checking Unique Usernames During Registration
# Scenario
# Prevent users from registering with an existing username.

registered_users = {"hari", "rahul", "priya"}

new_user = "hari"

if new_user in registered_users:
    print("Username already exists")
else:
    registered_users.add(new_user)
```

```python
# POC 2. Finding Common Skills Between Employees
# Scenario
# Find employees who share the same skills.

employee1 = {"Python", "SQL", "Django"}
employee2 = {"Python", "Java", "SQL"}

common_skills = employee1.intersection(employee2)

print(common_skills)
```

```python
# POC 3. Detecting Duplicate Email Addresses
# Scenario
# Identify duplicate email IDs from uploaded files.

emails = [
    "a@gmail.com",
    "b@gmail.com",
    "a@gmail.com"
]

if len(emails) != len(set(emails)):
    print("Duplicates found")

```

```python
# POC 4. Finding Missing Subjects
# Scenario
# Determine which mandatory subjects a student hasn't completed.

required = {"Math", "Science", "English"}
completed = {"Math", "English"}

missing = required - completed

print(missing)
```

```python
# POC 5. Permission Management
# Scenario
# Check if users have required permissions.

user_permissions = {"read", "write"}
required_permissions = {"read"}

if required_permissions.issubset(user_permissions):
    print("Access Granted")
```
