```python
# A loop in Python is a control flow statement used to repeatedly execute a block of code without rewriting it.
```

```python
# For Loop

# A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).
# This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-orientated programming languages.
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
```

```python
# Break With the break statement we can stop the loop before it has looped through all the items:

fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
    if x == "banana":
        break
```

```python
# Continue With the continue statement we can stop the current iteration of the loop, and continue with the next:

fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        continue
    print(x)
```

```python
# Range To loop through a set of code a specified number of times, we can use the range() function,
for x in range(6):
    print(x)
```

```python
for x in range(2, 6):
    print(x)
```

```python
for x in range(2, 30, 3):
    print(x)
```

```python
# Nested For

adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
    for y in fruits:
        print(x, y)
```

```python
# While Loop

# With the while loop we can execute a set of statements as long as a condition is true.
```

```python
i = 1
while i < 6:
    print(i)
    i += 1 
```

```python
# break 
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
# continue
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

```python
# Else

i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
```

```python
# POC 1. Display All Available Prompt Templates
# Scenario
# An AI system shows all available prompts.

templates = [
    "Summarize Text",
    "Translate Text",
    "Generate Quiz",
    "Create Blog"
]

for template in templates:
    print(template)
```

```python
# POC 2. Search for a Prompt
# Scenario
# Find whether a prompt exists.

templates = [
    "Email Generator",
    "Resume Reviewer",
    "Code Explainer"
]

search = input("Enter template name: ")

for template in templates:
    if template == search:
        print("Template Found")
        break
else:
    print("Template Not Found")
```

```python
# POC 3. Count Total Templates
# Scenario
# Count available prompts.

templates = [
    "Prompt1",
    "Prompt2",
    "Prompt3",
    "Prompt4"
]

count = 0

for template in templates:
    count += 1

print("Total Templates:", count)
```

```python
# POC 4. Filter AI Prompts
# Scenario
# Display only prompts containing "Generate".

templates = [
    "Generate Blog",
    "Summarize Article",
    "Generate Quiz",
    "Translate Text"
]

for template in templates:
    if "Generate" in template:
        print(template)
```

```python
# POC 5. User Menu Driven Prompt Manager
# Scenario
# Keep showing templates until user exits.

templates = [
    "Blog Generator",
    "Email Writer",
    "Code Reviewer"
]

while True:
    print("\nTemplates")

    for template in templates:
        print(template)

    choice = input("Exit? (yes/no): ")

    if choice == "yes":
        break
```

```python
# POC 6. Batch Prompt Processing
# Scenario
# Process prompts for multiple users.

users = ["Hari", "John", "Priya"]

template = "Generate Interview Questions"

for user in users:
    print(f"Running '{template}' for {user}")
```

```python
# POC 7. Prompt Usage Tracker
# Scenario
# Track how many times prompts were used.

used_prompts = [
    "Blog Generator",
    "Blog Generator",
    "Email Writer",
    "Blog Generator"
]

count = 0

for prompt in used_prompts:
    if prompt == "Blog Generator":
        count += 1

print("Used:", count, "times")
```

```python
# POC 8. Validate Prompt Templates
# Scenario
# Check for empty prompts.

templates = [
    "Generate Blog",
    "",
    "Translate Text",
    ""
]

for template in templates:
    if template == "":
        print("Invalid Template Found")
    else:
        print("Valid:", template)
```

```python
# 9. Smart Prompt Recommendation System
# Scenario
# Recommend prompts based on user skill.

templates = [
    "Python Basics",
    "Python OOP",
    "Python Advanced",
    "GenAI Projects"
]

skill = input("Enter skill level: ")

for template in templates:

    if skill == "beginner" and "Basics" in template:
        print(template)

    elif skill == "intermediate" and "OOP" in template:
        print(template)

    elif skill == "advanced" and ("Advanced" in template or "GenAI" in template):
        print(template)
```
