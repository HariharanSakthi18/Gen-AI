```python
#  List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.
#  newList = [expression for item in iterable if condition == True]
```

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newList = []
for x in fruits:
    if "a" in x:
        newList.append(x)

print(newList)
```

```python
# List comprehension 
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newList = [x for x in fruits if "a" in x]
print(newList)
```

```python
# Range of numbers
newList = [x for x in range(10)]
print(newList)

newList = [x for x in range(10) if x < 5]
print(newList)
```

```python
new = [x for x in range(10) if x%2==0 ]
print(new)
```

```python
# upper case
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newList = [x.upper() for x in fruits]
print(newList)
```

```python
#newList to 'hello'
newList = ['hello' for x in fruits]
print(newList)
```

```python
# Return "orange" instead of "banana":
newList = [x if x != "banana" else "orange" for x in fruits]
print(newList)
```

```python
a = [2, 3, 4, 5]
res = [val ** 2 for val in a]
print(res)
```

```python
a = [1, 2, 3, 4, 5]
res = [val for val in a if val % 2 == 0]
print(res)
```

```python
b = [5, 12, 7, 18, 3, 20]
res = [val for val in b if val > 10]
print(res)
```

```python
a = [i for i in range(10)]
print(a)
```

```python
c = [(x, y) for x in range(3) for y in range(3)]
print(c)
```

```python
mat = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
res = [val for row in mat for val in row]
print(res)
```

```python
prompts = [
    {"id": 1, "category": "email", "template": "Write a professional email to {name}."},
    {"id": 2, "category": "code", "template": "Explain the Python code: {code}"},
    {"id": 3, "category": "resume", "template": "Improve this resume summary."},
    {"id": 4, "category": "email", "template": "Generate a follow-up email."},
    {"id": 5, "category": "sql", "template": "Optimize this SQL query."}
]

# POC 1. Filter prompts by category
# Interview Scenario
# Show only email-related prompts.

email_prompts = [
    p for p in prompts
    if p["category"] == "email"
]

print(email_prompts)
```

```python
# POC 2. Extract only template texts
# Interview Question
# Get all prompt texts to display in UI.
templates = [p["template"] for p in prompts]

print(templates)
```

```python
# POC 3. Add status labels
# Real-Time GenAI Dashboard
# Mark prompts as active.
updated = [
    {**p, "status": "Active"}
    for p in prompts
]

print(updated)
```

```python
# POC 4. Find templates requiring variables
# Real-Time Use
# Detect prompts needing user input.
variable_prompts = [
    p for p in prompts
    if "{" in p["template"]
]

print(variable_prompts)
```

```python
# POC 5. Generate dropdown options
# Real-Time Frontend Use
# React/Angular applications often need this format.
dropdown = [
    {
        "label": p["template"],
        "value": p["id"]
    }
    for p in prompts
]

print(dropdown)
```
