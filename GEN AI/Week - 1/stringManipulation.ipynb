```python
# String manipulation means performing operations on strings such as accessing, modifying, searching, formatting, and transforming text
```

```python
# Creating a string
my_string = "Hello, World!"
print(my_string)
```

```python
# Accessing characters in a string
first_character = my_string[0]  # 'H'
last_character = my_string[-1]  # '!' 
print(first_character)
print(last_character)  
```

```python
# Modifying a string (strings are immutable, so we create a new string)
modified_string = my_string.replace("World", "Python") 
print(modified_string)  
```

```python
# Searching for a substring
substring_index = my_string.find("World")  
print(substring_index)
```

```python
# String formatting using format() method
myvar = {"name" : "Jane", "age" : 36}
var = "Happy birthday {name} you are now on level {age}!"
print(var.format_map(myvar))

txt1 = "For only {price:.2f} dollars!"
print(txt1.format(price = 49))
```

```python
# String {} then {} '.format{"value1, 'value2'}

# Formatting a string
name = "Alice"
greeting = f"Hello, {name}!"
print(greeting)

print('I am the trainer of {} {} {} ' .format('Besant', 'Technologies', 'Hebbal', 'Sahakarnagar'))

print('I am the trainer of {1} {0} {2}'.format('Besant', 'Technologies', 'Hebbal', 'Sahakarnagar', 'Kodigahelli'))

print('{t} I am the trainer of {t} {h} {b}'.format (h = 'Hebbal', t = 'Technologies', b = 'Besant'))
```

```python
# Float formatting follow " {value:width.precision f}" 
quo = 151/33
print(quo)

print("The quotient is {q:9.6f}".format(q=quo))


area = 'Hebbal'
pin = 560095
print('Besant Technologies, {}'.format(area))

print(f'{area} Besant Technologies, {pin}')

print(f'Besant Technologies, {area} {pin}')
```

```python
# POC 1: Basic Prompt Template Replacement
# Scenario
# Generate prompts dynamically.

template = "Write a {tone} email to {recipient} about {topic}."

prompt = template.format(
    tone="professional",
    recipient="HR",
    topic="leave request"
)

print(prompt)
```

```python
# POC 2: Store Multiple Templates Using Lists
# Scenario
# An AI application supports multiple prompt types.

templates = [
    "Summarize this text: {text}",
    "Translate {text} into {language}",
    "Generate interview questions for {role}"
]

print(templates[1])
```

```python
# POC 3: Search Templates by Keyword
# Scenario
# Find templates related to interviews.

templates = [
    "Generate interview questions for {role}",
    "Summarize {text}",
    "Write an email to {person}"
]

keyword = "interview"

for template in templates:
    if keyword.lower() in template.lower():
        print(template)
```

```python
# POC 4: Extract Variables from Templates
# Scenario
# Identify placeholders.

template = "Generate a {tone} email to {recipient}"

words = template.split()

variables = []

for word in words:
    if word.startswith("{") and word.endswith("}"):
        variables.append(word)

print(variables)
```

```python
# POC 5: Prompt Validation
# Scenario
# Check whether mandatory placeholders exist.

template = "Write an email to {recipient}"

required = "{recipient}"

if required in template:
    print("Valid Template")
else:
    print("Invalid Template")
```

```python
# POC 6: Prompt Versioning
# Scenario
# Maintain different versions.

versions = [
    "Summarize {text}",
    "Summarize {text} in bullet points",
    "Summarize {text} in simple English"
]

for i, prompt in enumerate(versions, start=1):
    print(f"Version {i}: {prompt}")
```

```python
# POC 7: Prompt Categorization
# Scenario
# Group prompts by category.

prompts = {
    "email": [
        "Write a follow-up email",
        "Write a resignation email"
    ],
    "interview": [
        "Generate Python questions",
        "Generate React questions"
    ]
}

print(prompts["interview"])
```

```python
# POC 8: Mini Prompt Template Manager
# Scenario
# Generate prompts from a selected template.

templates = [
    "Summarize {topic}",
    "Translate {topic} into {language}",
    "Generate interview questions on {topic}"
]

choice = 2

template = templates[choice]

prompt = template.format(
    topic="Python",
    language="Tamil"
)

print(prompt)
```
