```python
# Zip function
pairs = [("Apple", 5), ("Banana", 3), ("Orange", 2)]

# Split them back apart
fruits, quantities = zip(*pairs)
print(fruits)      # ('Apple', 'Banana', 'Orange')
print(quantities)
```

```python
# Example 
def aList(new):
    newList=[31,32,33,34,35]
    paired = list(zip(new,newList))
    combined = [item for pair in paired for item in pair]
    return combined

new = [1,2,3,4,5]
res = aList(new)
print(res)
```

```python
# KeyWord Arguments
def my_function(**kwargs): # kwargs saves input as dictionaries by default
    if 'fruit' in kwargs:
        print('My fruit of choice is {}'.format(kwargs['fruit']))
    else:
        print('I did not find any fruit here')
    
my_function(veg='tomato',car='BMW',animal='Tiger',fruit='apple')
```

```python
# LEGB Rules
name = 'This is an outer string'

def greet():
    name = 'Learning Again'
    print('Hi',name)

    def hello():
        name = 'New Learning'
        print('Hello',name)

        def demo():
            name = 'Learning'
            print('demo',name)

        demo()
            
    hello()
    
greet()
```

```python
# decorators
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Hello Sally"

print(myfunction())
```

```python
# Arguments in the Decorated Function
def changecase(func):
    def myinner(x):
        return func(x).upper()
    return myinner

@changecase
def myfunction(nam):
    return "Hello " + nam

print(myfunction("John"))
```

```python
# *args and **kwargs
def changecase(func):
    def myinner(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return myinner   

@changecase
def myfunction(name):
    return "Hello " + name

print(myfunction("John"))
```

```python
# POC 1: Display All Prompt Templates
# Real-Time Use
# Admin dashboard showing all available prompts.

prompt = [
    "Summarize this article",
    "Translate to French",
    "Generate Python code",
    "Write an email"
]

def display_prompts(prompt_list):
    for prompt in prompt_list:
        print(prompt)

display_prompts(prompt)
```

```python
# POC 2: Search Prompts by Category
# Real-Time Use
# Filter prompts in a GenAI application.

prompts = [
    {"title": "Professional Email", "category": "email"},
    {"title": "Meeting Reminder", "category": "email"},
    {"title": "Python Code Generator", "category": "coding"},
    {"title": "Story Writer", "category": "creative"}
]

def get_prompts_by_category(prompt_list, category):
    result = []

    for prompt in prompt_list:
        if prompt["category"] == category:
            result.append(prompt)

    return result

emails = get_prompts_by_category(prompts, "email")
print(emails)
```

```python
# POC 3: Delete a Prompt
# Real-Time Use
# Removing outdated prompts.

prompts = [
    {"id": 1, "title": "Email Generator", "category": "email"},
    {"id": 2, "title": "Python Tutor", "category": "coding"},
    {"id": 3, "title": "Story Writer", "category": "creative"}
]

def delete_prompt(prompt_list, prompt_id):
    for prompt in prompt_list:
        if prompt["id"] == prompt_id:
            prompt_list.remove(prompt)
            break

delete_prompt(prompts, 2)

print(prompts)
```

```python
# POC 4: Update Existing Prompt
# Real-Time Use
# Editing prompt templates.

def update_prompt(prompt_list, prompt_id, new_template):
    for prompt in prompt_list:
        if prompt["id"] == prompt_id:
            prompt["template"] = new_template
            return True

    return False

updated = update_prompt(
    prompts,
    1,
    "Write a professional email to {name}."
)

print(updated)
print(prompts)
```

```python
# POC 5: Count Prompts by Category
# Real-Time Use
# Analytics dashboard.

def count_by_category(prompt_list, category):
    count = 0

    for prompt in prompt_list:
        if prompt["category"] == category:
            count += 1

    return count

print(count_by_category(prompts, "email"))
```

```python
# POC 6: Add a New Prompt
# Real-Time Use
# Admin users creating new prompt templates.

def add_prompt(prompt_list, prompt_id, category, template):
    prompt = {
        "id": prompt_id,
        "category": category,
        "template": template
    }

    prompt_list.append(prompt)

add_prompt(
    prompts,
    4,
    "sql",
    "Optimize this SQL query."
)

print(prompts)
```

```python
# POC 7: Format Prompt with User Data
# Real-Time Use
# Generating final prompts before sending to LLMs.

def generate_prompt(template, data):
    return template.format(**data)

result = generate_prompt(
    "Write an email to {name}.",
    {"name": "Hariharan"}
)

print(result)
```

```python
# POC 8: Validate Prompt Variables
# Real-Time Use
# Check whether placeholders exist.

def has_variables(template):
    return "{" in template and "}" in template

print(has_variables("Write an email to {name}."))
print(has_variables("Hello World"))
```

```python
# POC 9: Generate Prompt Preview
# Real-Time Use
# UI cards displaying prompt summaries.

def get_preview(template, length=20):
    return template[:length]

print(
    get_preview(
        "Write a professional email to customer."
    )
)
```

```python
# POC 10: Find Prompt by ID
# Real-Time Use
# Selecting prompts from a dropdown.

def find_prompt(prompt_list, prompt_id):
    for prompt in prompt_list:
        if prompt["id"] == prompt_id:
            return prompt

    return None

result = find_prompt(prompts, 3)

print(result)
```

```python
# Decorators

def greet(func):
    def wrap(a,b):
        if a<b:
            a,b = b,a
        return func(a,b)
    return wrap

@greet
def div(a,b):
    return a/b

@greet
def sub(a,b):
    return a-b

result = div(2,4)
print(result)

result1 = sub(2,4)
print(result1)
```
