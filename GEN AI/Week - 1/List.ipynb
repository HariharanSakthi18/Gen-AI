```python
# List is a built-in data type in Python that is used to store multiple items in a single variable. 
# Lists are ordered, mutable (changeable), and allow duplicate elements. They are defined using square brackets 
# [] and can contain elements of different data types.
```

```python
# Creating a List
my_list = [1,2,3,"Hello",4.5,True]
print(my_list)
```

```python
# Accessing elements in a list
print(my_list[0])                           
print(my_list[3]) 
```

```python
# Modifying elements in a list
my_list[1] = 20                             
print(my_list) 
```

```python
# List operations
print(type(my_list))                        
print(len(my_list))

print("Add: ", my_list[0] + my_list[1])              
print("Sub: ", my_list[0] - my_list[1])              
print("Mul: ", my_list[0] * my_list[1])              
print("Div: ", my_list[1] / my_list[0])              
```

```python
# Slice

print(my_list[0:3])

print(my_list[-4:-2])

print(my_list[-2:-4])  # This will print an empty list because the start index is greater than the end index when using negative indexing, resulting in []
```

```python
# Concatenating two lists
new_list1 = [10, 2, 5]
new_list2 = [4, 3, 1]
print(new_list1 + new_list2) 
```

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
basket = ["Pen", "book", "bottle"]
val = fruits.pop(1)
print(val)                                 # This will print the removed element "banana"

basket.append(val)                        # This will add "banana" to the end of the basket list
print(basket) 
```

```python
# Sort
newList = ['A','h','r','a','c','b','z','Z','X']
numList = [6,8,9,1,5,4]
newList.sort()
numList.sort()
print(newList)                           
print(numList)
```

```python
# Reverse 
number = [1, 2, 3, 3, 4, 5]
number.reverse()                         
print(number)
```

```python
print(number.count(3))
```

```python
# Descending order
res = [5, 2, 9, 1, 5, 6]                
res.sort(reverse=True)
print(res)                              # This will print the sorted res list in descending order ([9, 6, 5, 5, 2, 1])
```

```python
#Swapping values in a list
a=5
b=6
a,b=b,a
print(a,b)
```

```python
# Find the largest number in a list 
numbers = [3,5,1,6,9,4]
first = 0
for n in numbers:
    if n > first:
        first = n
print(first)
```

```python
# Finding the second largest number in a list
first = second = float('-inf')

for n in numbers:
    if n > first:
        second = first
        first = n
    elif n > second and n!=first:
        second = n
print(second)
```

```python
# POC 1. Employee Salary Management
# Scenario
# A company stores employee salaries in a list.

salaries = [25000, 30000, 45000, 28000, 50000]

print("Highest Salary: ", max(salaries))
print("Lowest Salary: ", min(salaries))
print("Average Salary: ", sum(salaries) / len(salaries))
```

```python
# POC 2. E-commerce Cart
# Scenario
# Add products to a shopping cart.

cart = []

cart.append("Laptop")
cart.append("Mouse")
cart.append("Keyboard")

print("Cart Items:")
for item in cart:
    print(item)
```

```python
# POC 3. Find Duplicate Employees
# Scenario
# HR accidentally stored duplicate employee IDs.

emp_ids = [101, 102, 103, 101, 104, 102]

duplicates = []

for id in emp_ids:
    if emp_ids.count(id) > 1 and id not in duplicates:
        duplicates.append(id)

print("Duplicate IDs:", duplicates)
```

```python
# POC 4. Remove Duplicate Records
# Scenario
# Database records contain duplicates.

records = [1, 2, 3, 2, 4, 1, 5]

unique_records = []

for item in records:
    if item not in unique_records:
        unique_records.append(item)

print(unique_records)
```

```python
# POC 5. Online Voting System
# Scenario
# Count votes received by a candidate.

votes = ["A", "B", "A", "C", "A", "B"]

count_a = votes.count("A")

print("Votes for A:", count_a)
```

```python
# POC 6. Find Missing Number
# Scenario
# One number is missing in the sequence.

nums = [1, 2, 3, 5]

n = 5

expected = n * (n + 1) // 2
actual = sum(nums)

print("Missing Number:", expected - actual)
```
