```python
# Modules is a file containing definitions and statements. A module can define functions, classes and variables. 
# Modules help organize code into separate files so that programs become easier to maintain and reuse.
# instead of writing everything in one place, related functionality can be grouped into its own module and imported whenever needed.
```

```python
# Create a Module
def add(x, y):
    return (x+y)
```

```python
import Calculator as calc
print(calc.Add(10, 2))

# 1
from math import sqrt, factorial
print(sqrt(16))
print(factorial(6))
```

```python
# 2  Import All Names
from math import *
print(sqrt(16))
print(factorial(6))
```

```python
# 3   Import With Alias
import math as m
print(m.pi)
```

```python
# Types of Modules
# 1. Built-in Modules:

import random
print(random.randint(1, 5))

# 2. User-Defined Modules
import Calculator as calc
print(calc.Substract(20, 5))
```

```python
# 3. External (Third-Party) Modules:
import requests
r = requests.get("https://google.com")
print(r.status_code)
```

```python
# 4. Package Modules
from mypkg import utils      # mypkg is our own package 
print(utils.some_func())

def some_func():
    return "Hello"
```

```python
# Locating a Module
import sys
for p in sys.path:
    print(p)
```
