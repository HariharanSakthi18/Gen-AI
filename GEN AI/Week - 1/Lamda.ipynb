```python
# Lambda functions are small anonymous functions, A lambda function can take any number of arguments, but can only have one expression.

# Syntax 
#  lambda arguments : expression
```

```python
x = lambda a : a + 10
print(x(5))
```

```python
x = lambda a, b : a * b
print(x(5, 6))
```

```python
x = lambda a, b, c : a + b + c
print(x(5, 6, 2))
```

```python
a = 'GeeksforGeeks'
upper = lambda x: x.upper()  
print(upper(a))
```

```python
# Use Cases
# If else

check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check(5))   
print(check(-3))  
print(check(0))
```

```python
# Function
def myfunc(n):
    return lambda a : a * n

mydoubler = myfunc(2)

print(mydoubler(11))
```

```python
# Example 
def myfunc(n):
    return lambda a : a * n

mydoubler = myfunc(2)
mytripler = myfunc(3)

print(mydoubler(11))
print(mytripler(11))
```

```python
# List Comprehension
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())
```

```python
# Filter
c = [1, 2, 3, 4, 5, 6]
even = filter(lambda x: x % 2 == 0, c)
print(list(even))
```

```python
# Map
a = [1, 2, 3, 4]
double = map(lambda x: x * 2, a)
print(list(double))
```

```python
#reduce
from functools import reduce
a = [1, 2, 3, 4]
mul = reduce(lambda x, y: x * y, a)
print(mul)
```

```python
# Sorted
students = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
```

```python
# POC 1: Sort Prompts by Usage Count
# Real-Time Use
# Show the most frequently used prompts first.

prompts = [
    {"id": 1, "title": "Email", "usage": 15},
    {"id": 2, "title": "Blog", "usage": 8},
    {"id": 3, "title": "Summary", "usage": 20}
]
sorted_prompts = sorted(
    prompts,
    key=lambda p: p["usage"],
    reverse=True
)

print(sorted_prompts)
```

```python
# POC 2: Sort Prompts Alphabetically by Category
# Real-Time Use
# Admin dashboard organization.

prompts = [
    {"id": 1, "title": "Welcome Email", "category": "email"},
    {"id": 2, "title": "Code Review", "category": "coding"},
    {"id": 3, "title": "Blog Intro", "category": "writing"},
    {"id": 4, "title": "Follow-up Email", "category": "email"}
]
result = sorted(
    prompts,
    key=lambda p: p["category"]
)

print(result)
```

```python
# POC 3: Filter Email Prompts
# Real-Time Use
# Display prompts belonging to a specific module.
email_prompts = list(
    filter(
        lambda p: p["category"] == "email",
        prompts
    )
)

print(email_prompts)
```

```python
# POC 4: Extract Prompt Templates
# Real-Time Use
# Send only prompt texts to the frontend.

prompts = [
    {
        "id": 1,
        "title": "Welcome Email",
        "template": "Hello {name}, Welcome!"
    },
    {
        "id": 2,
        "title": "Blog",
        "template": "Write a blog about AI."
    },
    {
        "id": 3,
        "title": "Summary",
        "template": "Summarize the following text."
    }
]
templates = list(
    map(
        lambda p: p["template"],
        prompts
    )
)

print(templates)
```

```python
# POC 5: Convert Templates to Uppercase
# Real-Time Use
# Standardization before export.

upper_templates = list(
    map(
        lambda p: p["template"].upper(),
        prompts
    )
)

print(upper_templates)
```

```python
# POC 6: Find High Usage Prompts
# Real-Time Use
# Analytics dashboard.

prompts = [
    {"id": 1, "title": "Welcome Email", "usage": 15},
    {"id": 2, "title": "Blog Generator", "usage": 45},
    {"id": 3, "title": "Code Review", "usage": 32},
    {"id": 4, "title": "Summary", "usage": 20}
]

popular = list(
    filter(
        lambda p: p["usage"] > 30,
        prompts
    )
)

print(popular)
```

```python
# POC 7: Create Dropdown Options
# Real-Time Use
# Frontend dropdown generation.

prompts = [
    {
        "id": 1,
        "template": "Welcome Email",
        "usage": 20
    },
    {
        "id": 2,
        "template": "Blog Generator",
        "usage": 45
    },
    {
        "id": 3,
        "template": "Code Review",
        "usage": 30
    }
]

dropdown = list(
    map(
        lambda p: {
            "label": p["template"],
            "value": p["id"]
        },
        prompts
    )
)

print(dropdown)
```

```python
# POC 8: Check Whether Prompt Needs Variables
# Real-Time Use
# Identify prompts requiring user input.

requires_input = list(
    filter(
        lambda p: "{" in p["template"],
        prompts
    )
)

print(requires_input)
```

```python
# POC 9: Get Prompt Preview
# Real-Time Use
# Show a short description in cards.

preview = list(
    map(
        lambda p: p["template"][:15],
        prompts
    )
)

print(preview)
```

```python
# POC 10: Sort by Multiple Fields
# Real-Time Use
# Sort by category first, then usage.

prompts = [
    {"id": 1, "category": "email", "usage": 25},
    {"id": 2, "category": "coding", "usage": 40},
    {"id": 3, "category": "email", "usage": 60},
    {"id": 4, "category": "writing", "usage": 20},
    {"id": 5, "category": "coding", "usage": 15}
]

result = sorted(
    prompts,
    key=lambda p: (p["category"], -p["usage"])
)

print(result)
```
