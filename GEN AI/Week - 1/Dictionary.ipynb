```python
# Dictionaries are used to store data values in key:value pairs. A dictionary is a collection which is ordered, changeable and does not allow duplicates. 
# Dictionaries are written with curly brackets, and have keys and values
```

```python
# Create and print a dictionary:
dict = {"brand": "Ford","model": "Mustang", "year": 1964}
print(dict)
```

```python
# Accessing Items
print(dict["brand"])
```

```python
# duplicate values will overwrite existing values:
dict = {"brand": "Ford","model": "Mustang", "year": 1964, "year": 2020}
print(dict)
```

```python
print(len(dict))

print(type(dict))
```

```python
# Dictionary Items - Data Types
# The values in dictionary items can be of any data type:
d = {'k1':123, 'k2':[30,10,20], 'k3':{'insidekey':250, 'outsidekey':'Hi', 'new':[10,20,50]}} 
d['k3']['new'][2]
```

```python
# Addition of dictionaries
sum = d['k1'] + d['k3']['new'][1]
print(sum)
```

```python
print(d['k1'] + d['k3']['new'][1])  

print(d['k3'].items())
```

```python
# Dictionary Methods
new={'one':"a",'two':"b",'three':"BESANT"} 
print(new['one'].upper())
print(new['two'].lower())

print(new.keys())
print(new.values())
```

```python
#change the value of a specific item by referring to its key name:
n = new['two']="C"
print(n)
```

```python
#update() method will update the dictionary with the items from the given argument. The argument must be a dictionary, or an iterable of key:value pairs.
new.update({'four':"D"})
print(new)

#add a new item to the dictionary by using a new index key and assigning a value to it:
new['five'] = "E"
print(new)
```

```python
#remove the item with the specified key name:
new.pop('two')
print(new)

new.popitem() # removes the last inserted item
print(new)

del new['one'] # removes the item with the specified key name
print(new)

clear = new.clear() # empties the dictionary
print(clear)
```

```python
customer = {
    "Customer1" : {
        "ID":1,
        "Name":"Hari",
        "Salary":50000
    },
    "Customer2" : {
        "ID":2,
        "Name":"John",
        "Salary":40000
    }
}

print(customer)

#Concatenate two values in dictionary
AddCusName = customer["Customer1"]["Name"] + customer["Customer2"]["Name"]
print(AddCusName)

for x, obj in customer.items():
    print(x)

    for y in obj:
        print(y + ':', obj[y])
```

```python
# loop through a dictionary
dict = {"brand": "Ford","model": "Mustang", "year": 1964}
for x in dict:
    print(x) # prints the keys
```

```python
for x in dict:
    print(dict[x]) # prints the values
```

```python
for x in dict.values():
    print(x) # prints the values
```

```python
for x in dict.keys():
    print(x) # prints the keys
```

```python
for x, y in dict.items():
    print(x, y) # prints the keys and values
```

```python
#copy a dictionary
dict1 = dict.copy()
print(dict1)
```

```python
# POC 2. Word Frequency Counter
# Scenario
# Count occurrences of words.

sentence = "python is easy python is powerful"

frequency = {}

for word in sentence.split():
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
```

```python
# 9. Cricket Team Statistics (Dictionary)

# Store

# Player Name : Runs

# Tasks:

# Print

# Player who scored highest
# Total team runs
# Average runs

# Do not use shortcuts if you can solve it using loops.

highest_score = 0
totalruns = 0
highest_name = " "
average_score = 0

Players = {"Kohli": 100, "Dhoni": 80, "Rohit": 90, "Jadeja": 70}

for player, score in Players.items():
    totalruns += score
    
    if score > highest_score:
        highest_score = score
        highest_name = player
        
average_score = totalruns / len(Players)
print("Total Runs:", totalruns)
print(highest_name, "has the highest score of", highest_score)
print("Average Score:", average_score)
```
