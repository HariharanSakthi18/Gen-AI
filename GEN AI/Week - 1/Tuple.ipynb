```python
# Tuple is a collection of ordered values and it is a immutable so we can't change there values. 
# In Python tuples are written with parentheses.
```

```python
# Create a tuple and print it:
thisTuple = ("apple", "banana", "cherry")
print(thisTuple)
```

```python
print(thisTuple.count('apple'))                                              
print(thisTuple.index('banana'))                                              
print(thisTuple[1])
```

```python
# Tuple Operations

print(type(thisTuple))
print(len(thisTuple))

tup = (2,5,9,3,"hii","welcome",4.5)
print("Add: ", tup[0] + tup[1])
print("Sub: ", tup[0] - tup[1])
print("Mul: ", tup[0] * tup[1])
print("Div: ", tup[0] / tup[1])
```

```python
# Slice
print(tup[1:4])
print(tup[-4:-1])
```

```python
# Change Tuple Values
# Convert the tuple into a list to be able to change it:
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)
print(x)
```

```python
# Add items to a tuple
# Convert the tuple into a list to be able to add items:
a = ("apple", "banana", "cherry")
b = list(a)
b.append("orange")
a = tuple(b)
print(a)
```

```python
# Remove items from a tuple
# Convert the tuple into a list to be able to remove items:
c = ("apple", "banana", "cherry")
d = list(c)
d.remove("banana")
c = tuple(d)
print(c)
```

```python
# concatenate two tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
result = tuple1 + tuple2
print(result)
```

```python
# Repeating a tuple
tuple3 = ("Hello") * 3
print(tuple3) 
```

```python
# Unpacking a Tuple
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits
print(green)
print(yellow)
print(red)
```

```python
# Using Asterisk* for Unpacking
fruits = ("apple", "banana", "cherry", "orange", "kiwi")
(green, yellow, *red) = fruits
print(green)
print(yellow)
print(red)
```

```python
# POC 1. Database Records (Read-Only Data)
# Scenario:
# When fetching data from databases, records are often returned as tuples because they should not be modified accidentally.

employee = (101, "Hariharan", "Developer", 50000)

emp_id, name, role, salary = employee

print(emp_id)
print(name)
```

```python
# POC 4. Returning Multiple Values from Functions
# Scenario:
# Functions can return multiple values as tuples.

def calculate(a, b):
    return a + b, a - b

result = calculate(10, 5)

print(result)
```

```python
# POC 3. Days of the Week
# Scenario:
# Days are constant and should not be modified.

days = (
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
)

print(days[0])
```

```python
# POC 4. Dictionary Keys Using Tuples
# Scenario:
# Lists cannot be dictionary keys because they are mutable.

attendance = {
    (101, "2026-06-10"): "Present",
    (102, "2026-06-10"): "Absent"
}

print(attendance[(101, "2026-06-10")])
```
