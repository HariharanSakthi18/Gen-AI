```python
# Polymorphism means "many forms" and allows the same method, function or operator to behave differently depending on the object 
# or data it works with. This flexibility helps create more reusable, maintainable and scalable code.
```

```python
# Compile-time Polymorphism

class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num
        return result

# Create object
calc = Calculator()

# Using default arguments
print(calc.multiply())            
print(calc.multiply(4))           

# Using multiple arguments
print(calc.multiply(2, 3))       
print(calc.multiply(2, 3, 4))
```

```python
# Runtime Polymorphism (Overriding)

class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Polymorphic behavior
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Drive!")

class Boat:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Sail!")

class Plane:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
    x.move()
```

```python
# POC 1: Different Prompt Types
# Scenario
# An AI application supports different prompt types.

class Prompt:

    def generate_prompt(self):
        print("Generic Prompt")


class TextPrompt(Prompt):

    def generate_prompt(self):
        print("Summarize this article.")


class CodePrompt(Prompt):

    def generate_prompt(self):
        print("Write a Python function.")


class SQLPrompt(Prompt):

    def generate_prompt(self):
        print("Generate an SQL query.")


prompts = [
    TextPrompt(),
    CodePrompt(),
    SQLPrompt()
]

for prompt in prompts:
    prompt.generate_prompt()
```

```python
# POC 2: Prompt Preview Manager
# Scenario
# Before saving prompts, preview them.

class Prompt:
    def preview(self):
        print("Preview")
class Email_Prompt(Prompt):
    def preview(self):
        print("Email Prompt Preview")
class Resume_Prompt(Prompt):
    def preview(self):
        print("Resume Prompt Preview")
class Chat_Prompt(Prompt):
    def preview(self):
        print("Chat Prompt Preview")
        
Prompt_List = [
    Email_Prompt(),
    Resume_Prompt(),
    Chat_Prompt()
]

for prompt in Prompt_List:
    prompt.preview()
    
```

```python
# POC 3: Execute AI Prompts
# Scenario
# Every prompt executes differently.

class Prompt:

    def execute(self):
        print("Executing...")


class TranslationPrompt(Prompt):

    def execute(self):
        print("Translating text...")


class SummarizePrompt(Prompt):

    def execute(self):
        print("Summarizing document...")


class CodingPrompt(Prompt):

    def execute(self):
        print("Generating Python code...")


prompts = [
    TranslationPrompt(),
    SummarizePrompt(),
    CodingPrompt()
]

for prompt in prompts:
    prompt.execute()
```

```python
# POC 4: Prompt Validation
# Scenario
# Different prompts require different validation.

class Prompt:

    def validate(self):
        print("Validation")


class EmailPrompt(Prompt):

    def validate(self):
        print("Checking email format...")


class SQLPrompt(Prompt):

    def validate(self):
        print("Checking SQL syntax...")


class CodePrompt(Prompt):

    def validate(self):
        print("Checking Python syntax...")


prompts = [
    EmailPrompt(),
    SQLPrompt(),
    CodePrompt()
]

for prompt in prompts:
    prompt.validate()
```

```python
# POC 5: Prompt Export
# Scenario
# Export prompts into different formats.

class PromptExporter:

    def export(self):
        print("Export")


class JSONExporter(PromptExporter):

    def export(self):
        print("Exporting as JSON")


class CSVExporter(PromptExporter):

    def export(self):
        print("Exporting as CSV")


class PDFExporter(PromptExporter):

    def export(self):
        print("Exporting as PDF")


exporters = [
    JSONExporter(),
    CSVExporter(),
    PDFExporter()
]

for exporter in exporters:
    exporter.export()
```
