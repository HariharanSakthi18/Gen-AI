```python
# Inheritance allows us to define a class that inherits all the methods and properties from another class.
# Parent class is the class being inherited from, also called base class.
# Child class is the class that inherits from another class, also called derived class.
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def info(self):
        print("Animal Name:", self.name)
    
class Dog(Animal):
    def sound(self):
        print(self.name , "Bark")
        
d = Dog("Arnold")
d.info()
d.sound()
```

```python
# Super() function is used to call methods from a superclass following Python’s Method Resolution Order (MRO). 

class Animal:
    def __init__(self, name):
        self.name = name
        
    def info(self):
        print("Animal Name:", self.name)
        
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
        
    def details(self):
        print(self.name, "is a" , self.breed)
        
d = Dog("Buddy", "Golden Retriever")
d.info()
d.details()
```

```python
# POC 1: Basic Prompt Template Inheritance
# Scenario
# Every prompt has a name and category.
# Different prompt types inherit these common properties.

class Prompt:
    def __init__(self, name, category):
        self.name = name 
        self.category = category
        
    def showDetails(self):
        print(self.name, "-", self.category)
        
class EmailPrompt(Prompt):
    pass

class ResumePrompt(Prompt):
    pass

Prompt_List = [
    EmailPrompt("Welcome Email","Email"),
    ResumePrompt("Resume Generator", "Resume")
]

for prompt in Prompt_List:
    prompt.showDetails()
```

```python
# POC 2: Prompt Manager with Multiple Prompt Types
# Scenario
# Store different prompt objects in one list.

class Prompt:

    def __init__(self, title):
        self.title = title

    def display(self):
        print("Prompt:", self.title)


class SQLPrompt(Prompt):
    pass


class CodePrompt(Prompt):
    pass


manager = []

manager.append(SQLPrompt("Generate SQL Query"))
manager.append(CodePrompt("Generate Python Code"))

for prompt in manager:
    prompt.display()
```

```python
# POC 3: Prompt with Common Author Information
# Scenario
# Every prompt stores the creator's name.

class Prompt:

    def __init__(self, title, author):
        self.title = title
        self.author = author

    def show(self):
        print(self.title, "-", self.author)


class TranslationPrompt(Prompt):
    pass


class ChatPrompt(Prompt):
    pass


prompts = [
    TranslationPrompt("Translate to Tamil", "Hari"),
    ChatPrompt("Customer Support", "John")
]

for prompt in prompts:
    prompt.show()
```

```python
# POC 4: Prompt Usage Counter
# Scenario
# Every prompt tracks usage.

class Prompt:

    def __init__(self, title):
        self.title = title
        self.usage = 0

    def use(self):
        self.usage += 1

    def show_usage(self):
        print(self.title, "Used", self.usage, "times")


class EmailPrompt(Prompt):
    pass


prompt = EmailPrompt("Job Application")

prompt.use()
prompt.use()

prompt.show_usage()
```

```python
# POC 5: Prompt Status
# Scenario
# Every prompt has a status.

class Prompt:

    def __init__(self, title):
        self.title = title
        self.status = "Draft"

    def publish(self):
        self.status = "Published"

    def show(self):
        print(self.title, "-", self.status)


class ResumePrompt(Prompt):
    pass


prompt = ResumePrompt("Resume Builder")

prompt.publish()

prompt.show()
```
