```python
# Class
# A class is a collection of objects. Classes are blueprints for creating objects. 
# A class defines a set of attributes and methods that the created objects (instances) can have. 
```

```python
class Dog:
    species = "Canine"  # Class attribute

    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age  # Instance attribute
```

```python
# Object An Object is an instance of a Class. It represents the attributes and reflects the properties of an object.
```

```python
dog1 = Dog("Buddy", 3)
print(dog1.name) 
print(dog1.species)
```

```python
class Prompt:

    def __init__(self, prompt_id, title, category, text, usage):
        self.id = prompt_id
        self.title = title
        self.category = category
        self.text = text
        self.usage = usage


prompts = [
    Prompt(1, "Welcome Email", "Email", "Welcome user", 45),
    Prompt(2, "Bug Fix", "Coding", "Fix this bug", 90),
    Prompt(3, "Resume Review", "Career", "Review my resume", 25),
    Prompt(4, "SQL Generator", "Database", "Generate SQL", 65),
    Prompt(5, "Meeting Summary", "Office", "Summarize meeting", 55)
]
```

```python
# POC 1 — Display All Prompts
# Interview Question
# Display every prompt stored in the manager.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def show_all(self):

        for prompt in self.prompts:
            print(prompt.id)
            print(prompt.title)
            print(prompt.category)
            print(prompt.usage)
            print("-" * 20)


manager = PromptManager(prompts)
manager.show_all()
```

```python
# POC 2 — Search Prompt by ID
# Very common interview question.
class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def search(self, prompt_id):

        for prompt in self.prompts:

            if prompt.id == prompt_id:
                print(prompt.title)
                print(prompt.text)
                return

        print("Prompt Not Found")


manager = PromptManager(prompts)
manager.search(3)
```

```python
# POC 3 — Add New Prompt
# Asked in almost every CRUD interview.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def add_prompt(self, prompt):

        self.prompts.append(prompt)
        
    def show_all(self):   
        for prompt in self.prompts:
            print(prompt.title)


manager = PromptManager(prompts)

new_prompt = Prompt(
    6,
    "Invoice Generator",
    "Finance",
    "Create Invoice",
    10
)

manager.add_prompt(new_prompt)

manager.show_all()
```

```python
# POC 4 — Delete Prompt
# Frequently asked.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def delete_prompt(self, prompt_id):

        for prompt in self.prompts:

            if prompt.id == prompt_id:
                self.prompts.remove(prompt)
                break
            
    def show_all(self):

        for prompt in self.prompts:
            print(prompt.id)
            print(prompt.title)
            print("-" * 20)


manager = PromptManager(prompts)

manager.delete_prompt(2)

manager.show_all()
```

```python
# POC 5 — Update Usage Count
# Interview favorite.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def increase_usage(self, prompt_id):

        for prompt in self.prompts:

            if prompt.id == prompt_id:
                prompt.usage += 1
                print("Updated")
                
    def show_all(self):

        for prompt in self.prompts:
            print(prompt.id)
            print(prompt.title)
            print(prompt.category)
            print(prompt.usage)
            print("-" * 20)


manager = PromptManager(prompts)

manager.increase_usage(1)

manager.show_all()
```

```python
# POC 6 — Find Most Used Prompt
# Very common.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def most_used(self):

        highest = self.prompts[0]

        for prompt in self.prompts:

            if prompt.usage > highest.usage:
                highest = prompt

        print(highest.title)
        print(highest.usage)
        
    


manager = PromptManager(prompts)
manager.most_used()
```

```python
# POC 7 — Display Prompts Category Wise
# Common filtering question.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def by_category(self, category):

        for prompt in self.prompts:

            if prompt.category == category:
                print(prompt.title)


manager = PromptManager(prompts)

manager.by_category("Coding")
```

```python
# POC 8 — Count Prompts in Each Category
# Very common in interviews.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def category_count(self):

        result = {}

        for prompt in self.prompts:

            if prompt.category not in result:
                result[prompt.category] = 1
            else:
                result[prompt.category] += 1

        print(result)


manager = PromptManager(prompts)

manager.category_count()
```

```python
# POC 9 — Top 3 Most Used Prompts
# Frequently asked.

class PromptManager:

    def __init__(self, prompts):
        self.prompts = prompts

    def top_three(self):

        result = sorted(
            self.prompts,
            key=lambda prompt: prompt.usage,
            reverse=True
        )

        for prompt in result[:3]:
            print(prompt.title, prompt.usage)


manager = PromptManager(prompts)

manager.top_three()
```

```python
# POC 10 — Complete Menu Driven Prompt Manager
# Most realistic interview project.

class Prompt:

    def __init__(self, id, title, category, usage):
        self.id = id
        self.title = title
        self.category = category
        self.usage = usage


class PromptManager:

    def __init__(self):
        self.prompts = []

    def add(self, prompt):
        self.prompts.append(prompt)

    def display(self):
        for prompt in self.prompts:
            print(prompt.id, prompt.title, prompt.category, prompt.usage)

    def search(self, id):

        for prompt in self.prompts:

            if prompt.id == id:
                print(prompt.title)
                return

        print("Not Found")

    def delete(self, id):

        for prompt in self.prompts:

            if prompt.id == id:
                self.prompts.remove(prompt)
                print("Deleted")
                return

        print("Prompt Not Found")


manager = PromptManager()

manager.add(Prompt(1, "Welcome", "Email", 25))
manager.add(Prompt(2, "Bug Fix", "Coding", 80))

while True:

    print()
    print("1 Display")
    print("2 Search")
    print("3 Delete")
    print("4 Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        manager.display()

    elif choice == 2:
        id = int(input("Enter ID: "))
        manager.search(id)

    elif choice == 3:
        id = int(input("Enter ID: "))
        manager.delete(id)

    elif choice == 4:
        break
```
