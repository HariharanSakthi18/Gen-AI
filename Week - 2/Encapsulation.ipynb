```python
# The idea of encapsulation is to bind the data members and methods into a single unit.
# Helps in better maintainability, readability and usability as we do not need to explicitly pass data members to member methods
# Helps maintain data integrity by allowing validation and control over the values assigned to variables.
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name          # public attribute
        self.__salary = salary    # private attribute

emp = Employee("Fedrick", 50000)
print(emp.name)       
print(emp.__salary)
```

```python
# self.name = name: Public attribute, can be accessed directly.
# self.__salary = salary: Private attribute, cannot be accessed directly.
# print(emp.name): Prints "Fedrick" because name is public.
# print(emp.__salary): Raises an error because __salary is private and hidden.


# Why do we need Encapsulation?
# - Protects data from unauthorized access and accidental modification.
# - Controls data updates using getter/setter methods with validation.
# - Enhances modularity by hiding internal implementation details.
# - Simplifies maintenance through centralized data handling logic.
# - Reflects real-world scenarios like restricting direct access to a bank account balance.
```

```python
# Public Members

class Employee:
    def __init__(self, name):
        self.name = name   # public attribute

    def display_name(self):   # public method
        print(self.name)

emp = Employee("John")
emp.display_name()   # Accessible
print(emp.name)      # Accessible
```

```python
# Protected Members

class Employee:
    def __init__(self, name, age):
        self.name = name       # public
        self._age = age        # protected

class SubEmployee(Employee):
    def show_age(self):
        print("Age:", self._age)   # Accessible in subclass

emp = SubEmployee("Ross", 30)
print(emp.name)        # Public accessible
emp.show_age()   
```

```python
# Private Members

class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self.__salary = salary    # private

    def show_salary(self):
        print("Salary:", self.__salary)

emp = Employee("Robert", 60000)
print(emp.name)          # Public accessible
emp.show_salary()        # Accessing private correctly
# print(emp.__salary)    # Error: Not accessible directly
```

```python
# Getter and Setter Method 

class Student:
    def __init__(self, name):
        self.name = name
        self.__grade = 0

    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Grade must be between 0 and 100")

    def get_grade(self):
        return self.__grade

    def get_status(self):
        if self.__grade >= 60:
            return "Passed"
        else:
            return "Failed"

student = Student("Emil")
student.set_grade(85)
print(student.get_grade())
print(student.get_status())
```

```python
# POC 1: Add Prompt Templates
# Scenario
# An AI team stores reusable prompts.

class PromptManager:
    def __init__(self):
        self.__prompts = []
    def add_prompt(self, prompt):
        self.__prompts.append(prompt)
    def show_prompt(self):
        return self.__prompts
    
manager = PromptManager()

manager.add_prompt("Summarize this article")
manager.add_prompt("Translate to Tamil")

print(manager.show_prompt())
```

```python
# POC 2: Delete Prompt
# Scenario
# Remove outdated prompts.

class PromptManager:
    def __init__(self):
        self.__prompts = []
    def add_prompt(self, prompt):
        self.__prompts.append(prompt)
    def del_prompt(self, prompt):
        if prompt in self.__prompts:
            self.__prompts.remove(prompt)
    def show_prompt(self):
        return self.__prompts
    
manager = PromptManager()

manager.add_prompt("Translate to French")
manager.add_prompt("Generate SQL Query")

manager.del_prompt("Translate to French")

print(manager.show_prompt())
```

```python
# POC 3: Search Prompt
# Scenario
# Find a prompt by keyword.

class PromptManager:
    def __init__(self):
        self.__prompts = []

    def add_prompt(self, prompt):
        self.__prompts.append(prompt)

    def search_prompt(self, keyword):
        result = []

        for prompt in self.__prompts:
            if keyword.lower() in prompt.lower():
                result.append(prompt)

        return result


manager = PromptManager()

manager.add_prompt("Summarize article")
manager.add_prompt("Generate Python code")
manager.add_prompt("Generate SQL query")

print(manager.search_prompt("Generate"))
```

```python
# POC 4: Prevent Duplicate Prompts

class PromptManager:
    def __init__(self):
        self.__prompts = []
    def add_prompt(self, prompt):
        if prompt not in self.__prompts:
            self.__prompts.append(prompt)
        else:
            print("Prompt Already Exists")
    def show_prompt(self):
        return self.__prompts
    
manager = PromptManager()

manager.add_prompt("Hii")
manager.add_prompt("Hii")
manager.add_prompt("Good Morning")

print(manager.show_prompt())
```

```python
# POC 5: Update Existing Prompt
# Scenario
# AI team updates a prompt version.

class PromptManager:
    def __init__(self):
        self.__prompts = []
    def add_prompt(self, prompt):
        self.__prompts.append(prompt)
    def update_prompt(self, oldPrompt, newPrompt):
        if oldPrompt in self.__prompts:
            index = self.__prompts.index(oldPrompt)
            self.__prompts[index] = newPrompt
    def show_prompt(self):
        return self.__prompts
    
manager = PromptManager()

manager.add_prompt("Hii")
manager.update_prompt("Hii", "Hii Good Morning")

print(manager.show_prompt())
```
